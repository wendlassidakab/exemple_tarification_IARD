## Variables qualitatives

```python
df["Frequence"] = df["num_claims"] / df["earned_exposure"] 
```

```python
result = (
    df.groupby("insured_employment", as_index=False)
      .agg({
          "num_claims": "sum",
          "earned_exposure": "sum",
      })
)
result["Frequence/100maisons"] = result["num_claims"] / result["earned_exposure"] * 100
result
```

```python
def frequence_par_variable(df, var):
    """
    Calcule le nombre de sinistres, l'exposition et la fréquence
    pour chaque modalité d'une variable qualitative.
    """
    result = (
        df.groupby(var, as_index=False)
          .agg({
              "num_claims": "sum",
              "earned_exposure": "sum"
          })
    )
    result["% claims"] = result["num_claims"] / result["num_claims"].sum()
    result["% exposure"] = result["earned_exposure"] / result["earned_exposure"].sum()
    result["Frequence / 100 maisons"] = result["num_claims"] / result["earned_exposure"] * 100
    return result
```

```python
output_path = "Analyse_frequence.xlsx"

variables = ["client_start_year", "oldest_insured_year_birth", "insured_employment", "marital_status", "num_children", 
      "smoker", "policy_issue_year", "liability_limit", "region", "construction_year", "house_type", "floors_above_ground",
      "exterior_cladding", "heating_type", "auxiliary_heating", "alarm_system", "garage_type", "basement", "roof_type", "roof_year",
      "skylight_presence", "pool_presence", "water_body_proximity", "fire_station_proximity", "hydrant_proximity", "room_rental",
      "commercial_activities", "mortgage", "liability_limit", "deductible", "earthquake_coverage"]


writer = pd.ExcelWriter(output_path, engine="openpyxl")

for var in variables:
    result = frequence_par_variable(df, var)
    result.to_excel(writer, sheet_name=var[:31], index=False)

writer.close()
```
